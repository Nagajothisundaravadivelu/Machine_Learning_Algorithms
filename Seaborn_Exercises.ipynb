```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
sns.set_style('whitegrid')
```

```python
titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
sns.jointplot(x='fare', y='age', data=titanic)
```

```python
sns.histplot(titanic['fare'],bins=30,kde=False)
plt.show()
```

```python
sns.boxplot(x='class',y='age',data=titanic)
plt.show()
```

```python
sns.swarmplot(x='class',y='age',data=titanic,hue='sex')
plt.show()
```

```python
sns.countplot(x='sex',data=titanic)
plt.show()
```

```python
numeric_titantic=titanic.select_dtypes(include=['number'])
plt.figure(figsize=(10,7))
sns.heatmap(numeric_titantic.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
g=sns.FacetGrid(titanic, col='sex')
g.map(sns.histplot,'age')
```
