```python
#read the xl sheet

import pandas as pd
df=pd.read_excel('/content/Alarm Survey Data.xlsx')
print(df)
```

```python
#check the datatype  of each column
print(df.dtypes)
df.info()
```

```python
df.select_dtypes(include='object').columns
```

```python
df['alarm_rating']=pd.to_numeric(
    df['alarm_rating'].str.replace(' stars','')
)
print(df)
```

```python
#Missing data check
df.isna().sum()
```

```python
df[df.isna().any(axis=1)]
```

```python
df['sleep_quality']=df['sleep_quality'].fillna(df['sleep_quality'].median())
print(df)
```

```python
df['sleep_quality'].value_counts(dropna=False)
```

```python
#Inconsistent text & typos Check
#Find inconsistent values
df['activity_level'].value_counts(dropna=False)
```

```python
#fix inconsistent values
df['activity_level']=df['activity_level'].replace(
    {
        'light':'lightly active',
        'light_activity':'lightly active'

    }
)
df['activity_level'].value_counts()
```

```python
#Duplicate Data check
#Find duplicates
df[df.duplicated()]
```

```python
#remove duplicates
df=df.drop_duplicates()
#verify
df.duplicated().sum()
```
