```python
def category(balance):
  if balance>=100000:
    return "Premium"
  elif balance>=50000:
    return "Regular"
  else:
    return "Basic"
df['Customer_Category']=df['Balance'].apply(category)
print(df)
```

```python
def eligible(row):
  if row['Credit_Score']>=700 and row['Balance']>=30000:
    return "Eligible"
  else:
    return "Not Eligible"

df['Loan_Eligible']=df.apply(eligible,axis=1)
print(df)
```

```python
avg_balance=df.groupby('City')['Balance'].mean()
highest_City=avg_balance.idxmax()
print(highest_City)
count=df[
    (df['City']==highest_City)
    & (df['Credit_Score']>700)
    ].shape[0]
print(count)
```
